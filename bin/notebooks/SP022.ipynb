```python
#import libraries
import os
import pandas as pd
```

```python
#define paths 
project_path = "/Users/karenzhu/Desktop/Projects/MAIN_structure_project/work/SP001_050/SP022/"
ligT_blast_path = os.path.join(project_path, 'ligT_blast.m8')
psiblast_hits_path = os.path.join(project_path, 'psiblast_hits.fasta')
```

```python
#create a new file to store the modified headers and its sequences
output_file = open(os.path.join(project_path, "psiblast_hits.renamed.fasta"), "a")

psiblast_file = open(psiblast_hits_path, "r")

#read in contents, modify the headers, and output new headers and its contents
for line in psiblast_file:
    if line.startswith('>'):
        line = line.lstrip('>')
        full_acc_name = line.split(" ")[0]

        if "sp|" in full_acc_name:
            split_full_acc_name = full_acc_name.split("|")
            modified_acc_name = split_full_acc_name[1].split(".")[0]
            output_file.write('>' + modified_acc_name + "\n")

        if "pdb|" in full_acc_name:
            split_full_acc_name = full_acc_name.split("|")
            modified_acc_name = split_full_acc_name[1] + '_' + split_full_acc_name[2]
            output_file.write('>' + modified_acc_name + "\n")

        if "|" not in full_acc_name:
            modified_acc_name = full_acc_name.split(".")[0]
            output_file.write('>' + modified_acc_name + "\n")
    
    else:
        
        output_file.write(line)

#clost the output file connection after writing its contents
output_file.close()
 
```
